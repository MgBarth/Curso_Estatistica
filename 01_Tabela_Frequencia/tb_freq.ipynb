## Tabela de Frequências com Python

```python
import pandas as pd
import sqlalchemy

df = pd.read_csv("../data/points_tmw.csv", sep=',')

engine = sqlalchemy.create_engine("sqlite:///../data/tmw.db")

df.to_sql("points", engine, if_exists="replace", index=False)
```

#### Tab. de Frequências de produtos

Para fazer o acumulado das duas frequências, estamos utilizando o método '.cumsum()'.

```python
freq_produto = (df.groupby(["descProduto"])[["idTransacao"]]                                                # Frequência Absoluta
                  .count()
                  .rename(columns={"idTransacao" : "Frequência Absoluta"}))

freq_produto["Freq. Abs. Acum."] = freq_produto["Frequência Absoluta"].cumsum()                                     # Frequênia Absoluta Acumulada

freq_produto["Freq. Relativa (%)"] = freq_produto["Frequência Absoluta"] / freq_produto["Frequência Absoluta"].sum() * 100  # Frequência Relativa

freq_produto["Freq. Rel. Acum. (%)"] = freq_produto["Freq. Relativa (%)"].cumsum()                          # Frequência Relativa Acumulada

freq_produto
```

#### Tab. de Frequências de Categorias

```python
freq_cat = (df.groupby(df["descCategoriaProduto"])[["idTransacao"]]                                                 # Freq. Abs.
              .count()
              .rename(columns={"idTransacao" : "Frequência Absoluta"}))

freq_cat["Freq. Abs. Acumulada"] = freq_cat["Frequência Absoluta"].cumsum()                                         # Freq. Abs. Acumulada

freq_cat["Frequência Relativa (%)"] = freq_cat["Frequência Absoluta"] / freq_cat["Frequência Absoluta"].sum() * 100 # Frequência Relativa (%)

freq_cat["Freq. Rel. Acumulada (%)"] = freq_cat["Frequência Relativa (%)"].cumsum()                                 # Freq. Rel. Acumulada

freq_cat
```
